# Vision Transformer from Scratch on CIFAR-10

A compact implementation of patch embeddings, multi-head self-attention, pre-normalized encoder blocks, and classification with a learned class token.

## Patchification

A 32×32 RGB image is split into 4×4 patches. Each flattened patch is projected into the transformer embedding space and receives a learned positional embedding.

```python
import torch
from cifar10_vit.model import VisionTransformer

model = VisionTransformer(32, 4, 2, 4, 64, 128, 10)
logits = model(torch.randn(2, 3, 32, 32))
logits.shape
```

## Attention and encoder flow

Each encoder uses pre-layer normalization, vectorized scaled dot-product attention, an MLP with GELU, and two residual connections.
